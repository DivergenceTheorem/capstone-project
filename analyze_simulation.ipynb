```python
import random
import networkx as nx
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
import pandas as pd
```

```python
df_experiment_statistics = pd.read_csv("dbum_param_grid_results.csv")
df = df_experiment_statistics.copy()

param_grid_delta = [0.95, 0.65, 0.35]
param_grid_cost = [1, 0.9, 0.8, 0.7, 0.6, 0.5, 0.4, 0.3, 0.2, 0.1]
param_grid_radius = [3, 5, 8, 10]
```

```python
def tag_params(row):

    delta = row['delta']
    c = row['c']

    if delta == 0.95:
        row['delta_label'] = 'high_delta'
    elif delta == 0.65:
        row['delta_label'] = 'intermediate_delta'
    else:
        row['delta_label'] = 'low_delta'
    
    if c in [1, 0.9, 0.8, 0.7]:
        row['c_label'] = 'high_cost'
    elif c in [0.6, 0.5, 0.4]:
        row['c_label'] = 'intermediate_cost'
    elif c in [0.3, 0.2, 0.1]: 
        row['c_label'] = 'low_cost'

    return row
```

```python
def override_labels(row):

    labels = row['labels']

    lables_list = labels.split("|")

    if 'complete' in lables_list:
        row['labels'] = 'complete'

    return row
```

```python
df['delta_label'] = None
df['c_label'] = None

df = df.apply(tag_params, axis=1)
df = df.apply(override_labels, axis=1)
```

```python
df_radius_3 = df[df['radius'] == 3]
df_radius_5 = df[df['radius'] == 5]
df_radius_8 = df[df['radius'] == 8]
df_radius_10 = df[df['radius'] == 10]
```

```python
df_delta_high = df[df['delta_label'] == 'high_delta']
df_delta_intermediate = df[df['delta_label'] == 'intermediate_delta']
df_delta_low = df[df['delta_label'] == 'low_delta']
```

```python
df_cost_high = df[df['c_label'] == 'high_cost']
df_cost_intermediate = df[df['c_label'] == 'intermediate_cost']
df_cost_low = df[df['c_label'] == 'low_cost']
```

```python
def display_label_distribution(df):

    network_types_distr = list(df['labels'])
    network_types_distr_all = []

    for label in network_types_distr:

        if '|' in label:
            network_types = label.split('|')
            for type in network_types:
                network_types_distr_all.append(type)
        else:
            network_types_distr_all.append(label)

    print(pd.Series(network_types_distr_all).value_counts())
```

```python
for df_delta in [df_delta_high, df_delta_intermediate, df_delta_low]:
    print(f"Delta: {df_delta['delta_label'].iloc[0]}")
    display_label_distribution(df_delta)
    print("\n")
```

```python
for df_cost in [df_cost_high, df_cost_intermediate, df_cost_low]:
    print(f"Cost: {df_cost['c_label'].iloc[0]}")
    display_label_distribution(df_cost)
    print("\n")
```

```python
STRUCTURE_COLUMNS = {
    'complete': 'Complete',
    'star_like': 'Star-like',
    'weakly_star_like': 'Weakly Star-like',
    'regular': 'Regular',
    'nearly_regular': 'Near-regular',
    'empty': 'Empty',
    'other': 'Other',
}


def structure_table(df, group_col, row_labels, index_name):

    exploded = df.assign(label=df['labels'].str.split('|')).explode('label')

    unknown = set(exploded['label']) - set(STRUCTURE_COLUMNS)
    assert not unknown, f"Labels without a table column: {unknown}"

    table = pd.crosstab(exploded[group_col], exploded['label'])
    table = table.reindex(index=list(row_labels), columns=list(STRUCTURE_COLUMNS), fill_value=0)
    table = table.rename(index=row_labels, columns=STRUCTURE_COLUMNS)
    table.index.name = index_name
    table.columns.name = None

    return table
```

```python
# Table 1: Emerging network structures across different myopic radii
table_1 = structure_table(
    df,
    group_col='radius',
    row_labels={r: r for r in param_grid_radius},
    index_name='Myopic Radius',
)
table_1
```

```python
# Table 2: Emerging network structures across different decay parameters δ
table_2 = structure_table(
    df,
    group_col='delta_label',
    row_labels={'high_delta': 'High', 'intermediate_delta': 'Intermediate', 'low_delta': 'Low'},
    index_name='Decay parameter δ',
)
table_2
```

```python
# Table 3: Emerging network structures across different link costs c
table_3 = structure_table(
    df,
    group_col='c_label',
    row_labels={'high_cost': 'High', 'intermediate_cost': 'Intermediate', 'low_cost': 'Low'},
    index_name='Cost c',
)
table_3
```
